```python
df = df.drop_duplicates()
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
df_temp = df.copy()
```

```python
df_temp['Extracurricular Activities'] = df_temp['Extracurricular Activities'].map({"Yes":1, "No":0})
```

```python
df_temp
```

```python
#X = Input features
#y = Output features

X = df_temp.drop(columns=["Performance Index"]) #-> Input Feature.
y = df_temp['Performance Index'] #->Output Feature.
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
X_train
#train -  pattern recognise konse col target col ki accuracy ko increase kar raa hai
```

```python
X_test
```

Train linear Regression model


```python
from sklearn.linear_model import LinearRegression
```

```python
LinearModel = LinearRegression()
#Linearmodel -> object of class linear regression
```

```python
#Train the linearmodel on training set
LinearModel.fit(X_train,y_train)
```

prediction on the unseen data

```python
y_pred = LinearModel.predict(X_test)
```

##model evaluation







```python
#MSE,MAE,r2_score
from sklearn.metrics import mean_squared_error , mean_absolute_error , r2_score
```

```python
mean_squared_error(y_test,y_pred)
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
#testing accuracy
r2_score(y_test,y_pred)
```

```python
y_train_pred = LinearModel.predict(X_train)
```

```python
r2_score(y_train, y_train_pred)
```
